```python
## Recorro todos los desconectados

notandesconectados = []

for D in desconectados:
    deg = G.degree(D)
    sp = nx.shortest_path(G, source=SOURCE, target=D)
    nei = list(G.neighbors(D))
    #
    if deg > 2:
        print(f"AS: {D} | Spf({SOURCE}) {sp} | Len {len(sp)} | Deg {deg} | ")
        notandesconectados.append(D)
```

```python
## A ver que tienen arriba esos

for D in notandesconectados:
    nei = list(G.neighbors(D))
    print(f"AS: {D} - neighs: {nei}")
    print(f"--")
```
